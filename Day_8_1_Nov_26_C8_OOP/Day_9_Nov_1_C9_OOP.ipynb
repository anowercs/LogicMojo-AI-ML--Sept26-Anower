```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display_details(self):
        print("Student name: ", self.name)
        print("marks: ", self.marks)

    def check_result(self):
        if self.marks >= 40:
            print("Result: Pass")
        else:
            print("Result: Fail")

student1 = Student("Shafi", 95)
student2 = Student("Jasim", 73)

student1.display_details()
student2.display_details()

student1.check_result()
```

```python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def show_details(self):
        print("Employee name", self.name)
        print("Salary", self.salary)

class DataScientist(Employee):

    def work(self):
        print("Working on machine learning models")

emp1 = DataScientist("Ruhul", 8000)

emp1.show_details()
emp1.work()
        
```

```python
class Python:

    def python_skill(self):
        print("knows Python")


class SQL:
    def sql_skill(self):
        print("knows sql")


# a class can inherit more than one parent
class  DataScientist(Python, SQL):
    pass

person = DataScientist()

person.python_skill()

person.sql_skill()
        
```

```python
class Dog:
    def sound(self):
        print("Dog barks")

class Cat:
    def sound(self):
        print("cat meow")

class Cow:
    def sound(self):
        print("Cow Moos")


animals = [Dog(), Cat(), Cow()]

for animal in animals:
    animal.sound()
```
