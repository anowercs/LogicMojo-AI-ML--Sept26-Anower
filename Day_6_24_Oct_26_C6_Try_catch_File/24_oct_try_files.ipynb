```python
a = 10
b = 0


print(a/b)
```

```python
try:
    a = 10
    b = 0
    print(a/b)

except ZeroDivisionError:
    print("Cant devide by Zero")
```

```python
try:
    age = input("Pls enter the age: ")

    if age.strip() == "":
        raise ValueError("Input cant be empty")

    age = int(age)

    if age < 0:
        raise ValueError("Age cannot be negative")

    if age > 18:
        print("Eligible to vote")

    else:
        print("Not Eligible to vote")

except ValueError as e:
    print(e)
    
```

```python
try:
    num = int(input("Pls enter the number: "))

    if num <= 0:
        raise ValueError("The number is Negative or ZERO")

    if num > 0:
        if num % 2 == 0:
            print(" {num} is a even number.")
        else:
            print(" {num} is an odd number.")
    else:
        print("{num} is a negative or invalid number.")

except ValueError as e:
    print(e)
```
